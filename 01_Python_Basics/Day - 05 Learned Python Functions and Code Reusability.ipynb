```python
# Day 05 – Python Functions
# Overview
#Functions are reusable blocks of code that perform a specific task. Instead of writing the same code multiple times, functions allow us to write it once and use it whenever needed.
#Functions improve code readability, reduce duplication, and make programs easier to maintain.
# Learning Objectives
#By the end of this notebook, I will be able to:

#- Understand the purpose of functions
#- Create user-defined functions
#- Pass arguments to functions
#- Return values from functions
#- Use default and keyword arguments
#- Work with variable-length arguments
#- Apply lambda functions
```

```python
## 1. Creating Your First Function
#A function is created using the `def` keyword.
```

```python
def greet():
    print("Welcome to Python Functions!")

greet()
```

```python
#2. Function with Parameters
#Parameters allow us to pass information into a function.
```

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Diksha")
greet("Python Learner")
```

```python
#3. Function Returning a Value
#The `return` statement sends a result back to the caller.
```

```python
def square(number):
    return number ** 2

result = square(6)
print("Square =", result)
```

```python
#4. Multiple Parameters
#Functions can accept multiple inputs.
```

```python
def add(a, b):
    return a + b

print(add(15, 25))
```

```python
# 5. Default Parameters
#Default values are used when no argument is provided.
```

```python
def country(name="India"):
    print("Country:", name)

country()
country("Japan")
```

```python
#6. Keyword Arguments
#Arguments can be passed using parameter names.
```

```python
def student(name, age):
    print(name)
    print(age)

student(age=22, name="Diksha")
```

```python
#7. Variable-Length Arguments (*args)
#Use `*args` when the number of arguments is unknown.
```

```python
def total(*numbers):
    print(sum(numbers))

total(10, 20)
total(10, 20, 30, 40)
```

```python
#8. Keyword Variable Arguments (**kwargs)
#`**kwargs` accepts keyword arguments as a dictionary.
```

```python
def profile(**details):

    for key, value in details.items():
        print(key, ":", value)

profile(Name="Diksha", Course="MBA Business Analytics", City="Bhopal")
```

```python
#9. Lambda Functions
#Lambda functions are anonymous functions used for short operations.
```

```python
square = lambda x: x ** 2

print(square(9))
```

```python
# 10. Using Functions with Lists
```

```python
numbers = [10, 20, 30, 40, 50]

def average(data):
    return sum(data) / len(data)

print("Average =", average(numbers))
```

```python
#11. Practical Example
#Calculate the area of a rectangle.
```

```python
def rectangle_area(length, width):
    return length * width

print(rectangle_area(12, 5))
```

```python
#12. Mini Challenge
#Create a function that calculates the factorial of a number.
```

```python
def factorial(n):

    result = 1

    for i in range(1, n + 1):
        result *= i

    return result

print(factorial(5))
```
