```python
#1. Basic `for` Loop
#A `for` loop is used to iterate over a sequence such as a list, tuple, string, or range.
```

```python
# Basic for loop
for i in range(1, 6):
    print("Iteration:", i)
```

```python
#2. Looping Through a List
#Lists are iterable objects. We can access each element one by one using a `for` loop.
```

```python
fruits = ["Apple", "Banana", "Orange", "Mango"]

for fruit in fruits:
    print(fruit)
```

```python
## 3. Looping Through a String
#Python treats strings as iterable objects, allowing each character to be accessed individually.
```

```python
name = "Business Analytics"

for letter in name:
    print(letter)
```

```python
# 4. Using the `range()` Function
#The `range()` function generates a sequence of numbers and is commonly used with `for` loops.
```

```python
for number in range(10):
    print(number)
```

```python
# 5. `while` Loop
#A `while` loop executes repeatedly until the specified condition becomes False.
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# 6. `break` Statement
#The `break` statement immediately terminates the loop when a condition is met.
```

```python
for number in range(1, 11):

    if number == 6:
        break

    print(number)
```

```python
#7. `continue` Statement
#The `continue` statement skips the current iteration and proceeds to the next one.
```

```python
for number in range(1, 11):

    if number % 2 == 0:
        continue

    print(number)
```

```python
#8. Nested Loops
#Nested loops are useful when working with matrices, patterns, and multi-dimensional data.
```

```python
for i in range(3):

    for j in range(3):

        print(i, j)
```

```python
#9. Multiplication Table
#Loops can be used to generate multiplication tables efficiently.
```

```python
number = 7

for i in range(1, 11):
    print(f"{number} × {i} = {number * i}")
```

```python
#10. Factorial of a Number
#A factorial is the product of all positive integers less than or equal to a given number.
```

```python
number = 5
factorial = 1

for i in range(1, number + 1):
    factorial *= i

print("Factorial =", factorial)
```

```python
#11. Fibonacci Series
#The Fibonacci sequence is a classic programming exercise where each number is the sum of the previous two numbers.
```

```python
a = 0
b = 1

for i in range(10):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#12. Mini Challenge
#Print all even numbers between 1 and 50 using a loop.
```

```python
for number in range(1, 51):

    if number % 2 == 0:
        print(number)
```
